```python
import numpy as np
import pandas as pd
import plotly.express as px
import os
from sklearn.metrics import f1_score, accuracy_score, mean_squared_error, mean_absolute_error, r2_score, classification_report, roc_auc_score
import plotly.graph_objects as go
import plotly.figure_factory as ff
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.dummy import DummyClassifier
```

# Data loading

```python
# TODO: setup all the parameters

# dataset_name = "wesad3class"
#dataset_name = "prosi-lr"
dataset_name = "larfield"
# dataset_name = "prosi_subtasks"
# dataset_name = "wesad_3class_binary_tasks"

TASK_LIST = [
    "binary", "multiclass",
    "baseline-vs-non", "stress-vs-non", "amusement-vs-non",
    "baseline-vs-stress", "amusement-vs-stress", "amusement-vs-baseline",
    "3-running", "4-sitting"
]
task = 'gender'

classification_head = "ClassificationHeadMLP"

# get labels and dominant modalities for selected dataset/task
PROSI_CLASSES_LABELS = [
            "Sitting", "Paced breathing", "Sitting moving arm", "Sitting tapping",
            "Low pace", "Medium pace", "High pace"
        ]
# TODO: w labelkach wesada nie powinno być Neutral zamiast Baseline?
# TODO: co z dominant dla wesada? bo to nie do końca działa

if dataset_name == "prosi" or "prosi" in dataset_name:
    if task == "binary":
        classess_labels = ["Sitting", "Non-sitting"]
        # classess_labels = ["Stationary", "Non-stationary"]
        dominant_modality = "ACC"
    elif task == "multiclass":
        classess_labels = PROSI_CLASSES_LABELS[:]
        dominant_modality = "ACC"
    elif task == "3-running":
        classess_labels = ["Low pace", "Medium pace", "High pace"]
        dominant_modality = "ACC"
    elif task == "4-sitting":
        classess_labels = ["Sitting", "Paced breathing", "Sitting moving arm", "Sitting tapping"]
        dominant_modality = "ECG"
elif dataset_name == "wesad" or "wesad" in dataset_name:
    dominant_modality = "ECG"
    if task == "binary":
        classess_labels = ["Baseline", "Stress"]
    elif task == "multiclass":
        classess_labels = ["Baseline", "Stress", "Amusement", "Meditation"]
    if "3class" in dataset_name:
        if task == "multiclass":
            classess_labels = ["Baseline", "Stress", "Amusement"]
elif dataset_name == "larfield":
    if task == "gender":
        classess_labels = ["Female", "Male"]
        dominant_modality = "ECG"


num_classes = len(classess_labels)
print("dataset_name, task:", dataset_name, task)
print("classess_labels:", classess_labels)
```

```python
results = {}
# TODO: set experiment path
EXP_PATH = "/home/emognition/Desktop/multimodal-ssl/downstream_tasks/experiments_feature_learning"
logs_paths = [os.path.join(f"{EXP_PATH}/", lp) for lp in os.listdir(f"{EXP_PATH}/")]

for logs_path in logs_paths:
    if "all_subjects" in logs_path:
        continue
    if task in logs_path:
        results[logs_path.split("/")[-1].replace("pretrained", "initialized")] = {}
        for fold in os.listdir(logs_path):
            if classification_head in fold:
                results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold] = {}
                for result_file in os.listdir(os.path.join(logs_path, fold)):
                    if "predicted_labels" in result_file:
                        results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold]["predicted_labels"] = np.load(os.path.join(logs_path, fold, result_file)).squeeze()
                    elif "true_labels" in result_file:
                        results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold]["true_labels"] = np.load(os.path.join(logs_path, fold, result_file)).squeeze()
            
old_keys = list(results.keys())[:]
for key in old_keys:
    if "ACC_ECG" in key:        
        results[key.replace("ACC_ECG", "ECG_ACC")] = results.pop(key)
        print(key)
        
len(list(results.keys()))
```

# Count metrics

```python
for experiment, all_folds in results.items():
    all_true = []
    all_pred = []
    for fold, results_labels in all_folds.items():
        all_true += results_labels["true_labels"].tolist()
        all_pred += results_labels["predicted_labels"].tolist()
    # TODO: czy to dobrze że co iterację potencjalnie może być brana inna wartość dla pos_label dla zadania gender ??? czy to nie powinno być hardcoded
    results[experiment]["f1_score"] = f1_score(all_true,all_pred,  average= 'macro')      
    results[experiment]["accuracy"] = accuracy_score(all_true, all_pred)        
```

# Count baselines

```python
all_true_labels = []
all_true_labels = []
task_baselines = {}
for experiment, all_folds in results.items():
    for fold, results_labels in all_folds.items():
        if fold not in ["f1_score", "accuracy"]:
            all_true_labels += results_labels["true_labels"].tolist()
    break # only one experiment is needed; no matter which one
    
for dummy_strategy in ["most_frequent", "prior", "stratified", "uniform"]: # , "constant"
    clf = DummyClassifier(strategy=dummy_strategy, random_state=321)
    clf.fit(list(range(len(all_true_labels))), all_true_labels)
    dummy_preds = clf.predict(list(range(len(all_true_labels))))
    dummy_f1_score =  f1_score(y_true=all_true_labels, y_pred=dummy_preds, average='macro')        
    #dummy_accuracy =  accuracy_score(y_true=all_true_labels, y_pred=dummy_preds)
    task_baselines[dummy_strategy] = dummy_f1_score

task_baselines
```

## Aggregating results

```python
aggregated_results_mean = all_results.groupby(by=["experiment_description"])[["F1_score", "Accuracy"]].mean().add_suffix("_mean").reset_index()#.agg(['mean', 'std']).add_prefix()
aggregated_results_std = all_results.groupby(by=["experiment_description"])[["F1_score", "Accuracy"]].std().add_suffix("_std").reset_index()#.agg(['mean', 'std']).add_prefix()
aggregated_results_mean["lr"] = [e.split("lr_")[1].split("__")[0] for e in aggregated_results_mean["experiment_description"].values]
aggregated_results_mean
```

```python
aggregated_results = pd.merge(aggregated_results_mean, aggregated_results_std, on=["experiment_description"])
modalities = []
experiment_description_short = []

for ed in aggregated_results["experiment_description"]:
    if len(ed.split("__")) == 3:
        model_name, description, modality = ed.split("__")
    else:
        model_name_description, modality = ed.split("__")
        model_name_description = model_name_description.split("_")
        model_name, description = '_'.join(model_name_description[:-4]), '_'.join(model_name_description[-4:])
        model_name_description = None

    model_name = model_name.replace("unimodal_fusion", 'unimodal')

    for to_remove in ['initial_', 'ts_tcc_Larfield_', '_ecg', '_acc']:
        model_name = model_name.replace(to_remove, '')

    if description.split('_')[-1].endswith('classes'):
        description = '_'.join(description.split('_')[:-1])

    for to_remove in ['binary', 'multiclass', 'fusion_mean_', task]:
        description = description.replace(to_remove, '')

    description = description.replace('full_representation', 'FR')
    description = description.replace('feature_learning', 'FL')

    model_name = model_name.replace('full_representation', 'FR')
    model_name = model_name.replace('feature_learning', 'FL')
    
    if len(description) > 0 and description[0] != '_':
        description = '_' + description

    modalities.append(modality.replace('_test', ''))
    experiment_description_short.append(f"{model_name}{description}".split("__")[0].split("_2classes")[0].replace("_fix",""))

aggregated_results["modality"] = modalities
aggregated_results["experiment_description_short"] = experiment_description_short

list(aggregated_results['experiment_description_short'])
aggregated_results
```

## Gains

```python
backbones = set(all_results["experiment_description"].values)
backbones_multimodal = [b for b in backbones if "multimodal" in b]

# todo czy będą przypadki z 0 w mianowniku dla wariantu z komntarza
# todo: there will be cases with 0 for (metric_best - baseline)
def count_gain(result, baseline, metric_best=1):
    return (result-baseline) * 100 #/(metric_best - baseline)
```

```python
baseline_models = {
    # "ECG": "ts_tcc_Larfield_initial_unimodal_ecg__fusion_mean_binary_2classes__ECG_test",
    # "ACC": "ts_tcc_Larfield_initial_unimodal_acc__fusion_mean_binary_2classes__ACC_test",
    # "ECG_ACC":  'ts_tcc_Larfield_initial_unimodal_fusion__fusion_mean_binary_2classes__ECG_ACC_test',
    "ECG": f"ts_tcc_Larfield_initial_unimodal_ecg__fusion_mean_{task}_{num_classes}classes__ECG_test",
    "ACC": f"ts_tcc_Larfield_initial_unimodal_acc__fusion_mean_{task}_{num_classes}classes__ACC_test",
    "ECG_ACC":  f"ts_tcc_Larfield_initial_unimodal_fusion__fusion_mean_{task}_{num_classes}classes__ECG_ACC_test",
}

gains = {}
for bm in backbones_multimodal:
    gains[bm] = {}
    for modality in ["ECG", "ACC", "ECG_ACC"]:
        if bm.split("__")[-1].split("_test")[0] == modality:
            gains_temp_f1_score, gains_temp_accuracy = [], []
            for fold in folds:
                gains_temp_f1_score.append(count_gain(results[bm][fold]["f1_score"], results[baseline_models[modality]][fold]["f1_score"]))
                gains_temp_accuracy.append(count_gain(results[bm][fold]["accuracy"], results[baseline_models[modality]][fold]["accuracy"]))

            gains[bm][modality] = {
                "F1_mean": np.mean(gains_temp_f1_score),
                "F1_std": np.std(gains_temp_f1_score),
                "accuracy_mean": np.mean(gains_temp_accuracy),
                "accuracy_std": np.std(gains_temp_accuracy)
            }
```

```python
f1_score_gains_means, f1_scores_gains_std = [], []
accuracies_gains_means, accuracies_gains_stds = [], []
backbones, modalities = [], []

for backbone, modalities_results in gains.items():
    for modality, gains_results in modalities_results.items():
        f1_score_gains_means.append(gains[backbone][modality]["F1_mean"])
        f1_scores_gains_std.append(gains[backbone][modality]["F1_std"])
        accuracies_gains_means.append(gains[backbone][modality]["accuracy_mean"])
        accuracies_gains_stds.append(gains[backbone][modality]["accuracy_std"])

        backbone = backbone.split("_initial_")[1]
        backbone = backbone.replace("feature_learning", "FL").replace("full_representation", "FR").replace("__fusion", "_fusion").split("_fusion")[0]
        
        backbones.append(backbone)
        modalities.append(modality)

all_results_gains = pd.DataFrame()
all_results_gains["backbone"] = [b for b in backbones]
all_results_gains["modality"] = modalities
all_results_gains["F1_gain_mean"] = f1_score_gains_means
all_results_gains["F1_gain_std"] = f1_scores_gains_std
all_results_gains["accuracy_gain_mean"] = accuracies_gains_means
all_results_gains["accuracy_gain_std"] = accuracies_gains_stds

sorted(backbones)
```
