# Data loading

```python
import numpy as np
import pandas as pd
import plotly.express as px
import os
from sklearn.metrics import f1_score, accuracy_score, mean_squared_error, mean_absolute_error, r2_score, classification_report, roc_auc_score
import plotly.graph_objects as go
import plotly.figure_factory as ff
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
from decimal import Decimal, ROUND_HALF_UP


def load_dataset_task(dataset_name, task, classification_head="ClassificationHeadMLP"):
    results = {}
    logs_paths = [os.path.join(f"{EXPERIMENTS_PATH}/{dataset_name}/logs_test_pred/", lp)
                  for lp in os.listdir(f"{EXPERIMENTS_PATH}/{dataset_name}/logs_test_pred/")]
    for logs_path in logs_paths:
        if "all_subjects" in logs_path:
            continue
        if task in logs_path:
            results[logs_path.split("/")[-1].replace("pretrained", "initialized")] = {}
            for fold in os.listdir(logs_path):
                if classification_head in fold:
                    results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold] = {}
                    for result_file in os.listdir(os.path.join(logs_path, fold)):
                        if "predicted_labels" in result_file:
                            results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold]["predicted_labels"] = np.load(os.path.join(logs_path, fold, result_file)).squeeze()
                        elif "true_labels" in result_file:
                            results[logs_path.split("/")[-1].replace("pretrained", "initialized")][fold]["true_labels"] = np.load(os.path.join(logs_path, fold, result_file)).squeeze()
    old_keys = list(results.keys())[:]
    for key in old_keys:
        if "ACC_ECG" in key:
            results[key.replace("ACC_ECG", "ECG_ACC")] = results.pop(key)
    return results


# TODO: set local path with experiments
EXPERIMENTS_PATH = "D:/Projects/emoai/multimodal-ssl/downstream_tasks/experiments"
DATASETS = [
    # 'prosi-lr'
    # 'prosi_subtasks',
    # 'wesad3class',
    'wesad_3class_binary_tasks',
]

# example tesks
TASKS = [
#     "binary", "multiclass",
#     "3-running", "4-sitting"
    "baseline-vs-non", "amusement-vs-non", # "stress-vs-non", 
#     "baseline-vs-stress", "amusement-vs-stress", "amusement-vs-baseline",
]

all_datasets_results = {}
for dataset in DATASETS:
    for task in TASKS:
        all_datasets_results[f"{dataset}__{task}"] = load_dataset_task(dataset, task)
print(list(all_datasets_results.keys()))
```

# Count metrics

```python
for dataset_task, results in all_datasets_results.items():
    for experiment, all_folds in results.items():
        for fold, results_labels in all_folds.items():
                all_folds[fold]["f1_score"] = f1_score(results_labels["true_labels"], results_labels["predicted_labels"], pos_label=results_labels["true_labels"][0] if "gender" in task else 1,
                                                  average= 'binary' if '2classes' in task else 'macro')        
                all_folds[fold]["accuracy"] = accuracy_score(results_labels["true_labels"], results_labels["predicted_labels"])
```

# Visualize metrics

```python
all_datasets_metrics = {}

for dataset_task, results in all_datasets_results.items():
    f1_scores, accuracies, aucs, experiment_descriptions = [], [], [], []
    backbones, folds, modalities, tasks = [], [], [], []

    for experiment, all_folds in results.items():
        for fold, results_labels in all_folds.items():
            f1_scores.append(all_folds[fold]["f1_score"])
            accuracies.append(all_folds[fold]["accuracy"])
            experiment_descriptions.append(experiment)
            folds.append(fold)
    
    all_results = pd.DataFrame()
    all_results["Accuracy"] = accuracies
    all_results["F1_score"] = f1_scores
    all_results["fold"] = folds
    all_results["experiment_description"] = experiment_descriptions
    
    modalities = []
    for ed in experiment_descriptions:
        try:
            modalities.append(ed.split("__")[1].split("_test")[0])
        except:
            modalities.append("ECG")
    all_results["modality"] = modalities    
    all_datasets_metrics[dataset_task] = all_results

all_results
```

## Aggregating results

```python
all_datasets_metrics_agg = {}
for dataset_task, all_results in all_datasets_metrics.items():
    aggregated_results_mean = all_results.groupby(by=["experiment_description"])[["F1_score", "Accuracy"]].mean().add_suffix("_mean").reset_index()#.agg(['mean', 'std']).add_prefix()
    aggregated_results_std = all_results.groupby(by=["experiment_description"])[["F1_score", "Accuracy"]].std().add_suffix("_std").reset_index()#.agg(['mean', 'std']).add_prefix()
    
    aggregated_results = pd.merge(aggregated_results_mean, aggregated_results_std, on=["experiment_description"])
    modalities, experiment_description_short = [], []
    
    for ed in aggregated_results["experiment_description"]:
        if len(ed.split("__")) == 3:
            model_name, description, modality = ed.split("__")
        else:
            model_name_description, modality = ed.split("__")
            model_name_description = model_name_description.split("_")
            model_name, description = '_'.join(model_name_description[:-4]), '_'.join(model_name_description[-4:])
            model_name_description = None
    
        model_name = model_name.replace("unimodal_fusion", 'unimodal')
    
        for to_remove in ['initial_', 'ts_tcc_Larfield_', '_ecg', '_acc']:
            model_name = model_name.replace(to_remove, '')
    
        if description.split('_')[-1].endswith('classes'):
            description = '_'.join(description.split('_')[:-1])

        for to_remove in ['binary', 'multiclass', 'fusion_mean_', *TASKS]:
            description = description.replace(to_remove, '')
    
        description = description.replace('full_representation', 'FR')
        description = description.replace('feature_learning', 'FL')
    
        model_name = model_name.replace('full_representation', 'FR')
        model_name = model_name.replace('feature_learning', 'FL')
        
        if len(description) > 0 and description[0] != '_':
            description = '_' + description
    
        modalities.append(modality.replace('_test', ''))
        experiment_description_short.append(f"{model_name}{description}")

    aggregated_results["modality"] = modalities
    aggregated_results["experiment_description_short"] = experiment_description_short
    aggregated_results = aggregated_results.sort_values(by="modality")
    
    all_datasets_metrics_agg[dataset_task] = aggregated_results
```

## Generate table

```python
model_variants = [
    "unimodal",
    "multimodal",
    "multimodal_initialized_encoders",
    "multimodal_FR_initialized_encoders",
    "multimodal_FL_initialized_encoders",
    "multimodal_FL_1_1_initialized_encoders",
]

all_setups = list(all_datasets_metrics_agg.keys())
all_setups = [
    x.replace('__', '').replace('_', ' ').replace('subtasks', '')
    for x in all_setups
]
print("\\begin{table*}[t]")
print("    \centering")
print("    \\resizebox{\\textwidth}{!}{")
print("    \\begin{tabular}{lccc ccc}")
print("        \\toprule")

datasets_row, modalities_row = "        ", "        \\textbf{Representation Model}"
for dataset_task in all_setups:
    datasets_row +=  " & \multicolumn{3}{c}{\\textbf{" + dataset_task + "}}"
    modalities_row += " & \\textbf{ACC} & \\textbf{ECG} & \\textbf{Fusion}"
print(datasets_row + "\\\ \cmidrule(lr){2-4} \cmidrule(lr){5-7}")
print(modalities_row + "\\\ \midrule")

for experiment_description_short in model_variants:
    exp_variant = experiment_description_short.replace('_', ' ')
    exp_variant = exp_variant.replace(' initialized_encoders', '')
    latex_line = "        " + str(exp_variant)
    exp_results = ""
    for dataset_task, aggregated_results in all_datasets_metrics_agg.items():
        experiment_results_df = aggregated_results[aggregated_results["experiment_description_short"]==experiment_description_short]
        if experiment_results_df.empty:
            print("empty df for: ", experiment_description_short)
            continue
        f_modality, std_modality,  = {}, {}
        for modality in ['ACC', 'ECG', 'ECG_ACC']:
            experiment_results_modality_df = experiment_results_df[experiment_results_df["modality"]==modality]
            tmp_mean = Decimal(experiment_results_modality_df.F1_score_mean.values.tolist()[0])
            tmp_std = Decimal(experiment_results_modality_df.F1_score_std.values.tolist()[0])
            
            f_modality[modality] = tmp_mean.quantize(Decimal('0.001'), rounding=ROUND_HALF_UP)
            std_modality[modality] = tmp_std.quantize(Decimal('0.001'), rounding=ROUND_HALF_UP)
            exp_results += " & " + str(f_modality[modality]) + " $\pm$ " + str(std_modality[modality])
    
    latex_line += exp_results + " \\\ "
    print(latex_line)

print("        \\bottomrule")
print("    \end{tabular}}")
print("    \caption{Results}")
print("    \label{tab:results_prosi}")
print("\end{table*}")
```
